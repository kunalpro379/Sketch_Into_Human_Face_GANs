```python
import os
import cv2
import numpy as np

# Define directories
base_dir = os.path.dirname(os.getcwd())
final_dataset = os.path.join(base_dir, "DataPreprocessing", "finaldataset")
real_dir = os.path.join(final_dataset, "real")
sketch_dir = os.path.join(final_dataset, "sketches")
output_dir = os.path.join(base_dir, "preprocessedImages", "concatenated")

# Create output directory if not exists
os.makedirs(output_dir, exist_ok=True)

# Get image filenames (assuming real and sketch images have the same names)
real_images = set(os.listdir(real_dir))
sketch_images = set(os.listdir(sketch_dir))
common_images = real_images.intersection(sketch_images)

for image_name in common_images:
    real_path = os.path.join(real_dir, image_name)
    sketch_path = os.path.join(sketch_dir, image_name)
    output_path = os.path.join(output_dir, image_name)

    # Read images
    real_img = cv2.imread(real_path)
    sketch_img = cv2.imread(sketch_path)

    # Ensure both images exist and have the same height
    if real_img is None or sketch_img is None:
        print(f"Skipping {image_name} due to missing file.")
        continue

    # Resize the images to have the same height (optional)
    if real_img.shape[0] != sketch_img.shape[0]:
        height = min(real_img.shape[0], sketch_img.shape[0])
        real_img = cv2.resize(real_img, (int(real_img.shape[1] * height / real_img.shape[0]), height))
        sketch_img = cv2.resize(sketch_img, (int(sketch_img.shape[1] * height / sketch_img.shape[0]), height))

    # Concatenate horizontally
    concatenated = np.hstack((real_img, sketch_img))

    # Save the image
    cv2.imwrite(output_path, concatenated)
    print(f"Saved: {output_path}")

print("Processing complete.")
```
