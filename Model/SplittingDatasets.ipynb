```python
# Install and load required packages
if (!require("fs")) install.packages("fs")  # For better file operations
library(fs)

# Set random seed for reproducibility
set.seed(42)

# Print hello
cat("hello\n")

# Define paths using forward slashes
base_dir <- getwd()
split_dir <- file.path(base_dir, "testing_dataset_splitted")
left_dir <- file.path(split_dir, "left_split")
right_dir <- file.path(split_dir, "right_split")

# Create train-test directories
train_dir <- file.path(base_dir, "train")
test_dir <- file.path(base_dir, "test")

# Create subdirectories for sketches and real faces
train_sketch_dir <- file.path(train_dir, "sketches")
train_real_dir <- file.path(train_dir, "real")
test_sketch_dir <- file.path(test_dir, "sketches")
test_real_dir <- file.path(test_dir, "real")

# Create all directories
directories <- c(train_dir, test_dir, train_sketch_dir, train_real_dir, 
                test_sketch_dir, test_real_dir)
lapply(directories, dir_create, recurse = TRUE)

# Get all image numbers
image_files <- dir_ls(left_dir, regexp = "_left_split\\.jpg$")
image_numbers <- as.numeric(gsub("_left_split\\.jpg$", "", basename(image_files)))

# Shuffle and split
image_numbers <- sample(image_numbers)  # Shuffle
split_idx <- floor(length(image_numbers) * 0.8)  # 80-20 split
train_numbers <- image_numbers[1:split_idx]
test_numbers <- image_numbers[(split_idx + 1):length(image_numbers)]

# Function to copy images
copy_images <- function(numbers, sketch_dir, real_dir) {
  for (num in numbers) {
    # Define source and destination paths
    src_sketch <- file.path(left_dir, paste0(num, "_left_split.jpg"))
    dst_sketch <- file.path(sketch_dir, paste0(num, ".jpg"))
    
    src_real <- file.path(right_dir, paste0(num, "_right_split.jpg"))
    dst_real <- file.path(real_dir, paste0(num, ".jpg"))
    
    # Copy files if they exist
    if (file_exists(src_sketch) && file_exists(src_real)) {
      file_copy(src_sketch, dst_sketch, overwrite = TRUE)
      file_copy(src_real, dst_real, overwrite = TRUE)
    }
  }
}

# Copy files to train and test directories
cat("Copying training images...\n")
copy_images(train_numbers, train_sketch_dir, train_real_dir)
cat("Training set:", length(train_numbers), "images\n")

cat("Copying testing images...\n")
copy_images(test_numbers, test_sketch_dir, test_real_dir)
cat("Testing set:", length(test_numbers), "images\n")

cat("Dataset splitting complete!\n")
```
